```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model

llm =init_chat_model("google_genai:gemini-2.5-flash-lite", temperature=0)

response = llm.invoke(" Hola soy David y mi telefono es 317 123 4567")
response.pretty_print()
```

```python
from pydantic import BaseModel, Field

class ContactInfo(BaseModel):
    nombre : str = Field(description="El nombre de la persona")
    email : str = Field(description="El correo electrónico de la persona")
    telefono : str = Field(description="El número de teléfono de la persona")
    tono : int = Field(description="El tono en el que la persona", ge=0,le=100)
    años : int = Field(description="Los años de la persona")
    sentimiento : str = Field(description="El sentimiento de la persona, puede ser positivo, negativo o neutral")
    
llm_with_structured_output = llm.with_structured_output(schema=ContactInfo)    

messages = [
    ("system", "Eres un asistente que extrae información de contacto de personas a partir de textos."),
    ("user", " Hola soy David, mi telefono es 317 123 4567"),
]

response = llm_with_structured_output.invoke(messages)
response
```
