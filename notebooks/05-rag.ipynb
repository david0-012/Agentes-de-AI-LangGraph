```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("openai:gpt-5-nano", temperature=0)
```

```python
file_search_tool = {
    "type": "file_search",
    "vector_store_ids": ["vs_690bab2297a0819185301cbc12831ede"],
}

ll_with_tools = llm.bind_tools([file_search_tool])
```

```python
response = ll_with_tools.invoke("Como mejoro el rendimiento de mi website?")
response.text
```
